## GDP Development Analysis (2000–2022)

The figure compares real GDP across seven countries from 2000 to 2022, using constant US dollars.

### Main Findings

**United States:** The US remained the largest economy in the sample, with substantial long-term GDP growth. However, economic output declined during the 2008–2009 financial crisis and the COVID-19 pandemic in 2020.

**China:** China experienced the fastest and most sustained expansion among these countries. Its real GDP increased dramatically, narrowing the gap with the United States.

**Japan:** Japan experienced relatively slow economic growth, with its GDP remaining broadly stable compared with China's rapid expansion.

**Germany:** Germany showed moderate long-term growth, interrupted by economic downturns, particularly during the global financial crisis and the pandemic.

**United Kingdom:** The UK experienced overall GDP growth, with significant setbacks during the financial crisis and in 2020.

**Brazil:** Brazil experienced uneven GDP growth. After a period of expansion, economic performance weakened in the mid-2010s, followed by a further decline in 2020.

**Switzerland:** Switzerland experienced relatively steady long-term growth, although its total GDP remained smaller than that of the larger economies.

### Conclusion

Overall, China demonstrated the strongest long-term expansion, while the United States maintained the highest real GDP level. The results also show that economic crises affected multiple countries.

These comparisons reflect aggregate real GDP rather than GDP per capita, and therefore do not directly measure living standards.
